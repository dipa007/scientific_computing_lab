**Initialization and Helper Functions**

```python
import math

# Function for Q1 and Q2: f(x) = 1 / (1 + x^2)
def f_q1_q2(x):
    return 1.0 / (1.0 + x**2)

# Function for Q3: f(x) = sqrt(1 - x^2) - x
def f_q3(x):
    return math.sqrt(1.0 - x**2) - x

# True values for error computation
# Integral of 1/(1+x^2) from 0 to 1 is arctan(1) = pi/4
TRUE_VAL_Q1_Q2 = math.pi / 4.0

# Integral of sqrt(1-x^2) - x from 0 to 1/sqrt(2) is analytically pi/8
TRUE_VAL_Q3 = math.pi / 8.0
```

Question 1 - Computing $\pi$ via Trapezoidal and Simpson's Rules

```python
def trapezoidal(f, a, b, n):
    h = (b - a) / n
    s = 0.5 * (f(a) + f(b))
    for i in range(1, n):
        s += f(a + i * h)
    return s * h

def simpson_13(f, a, b, n):
    h = (b - a) / n
    s = f(a) + f(b)
    for i in range(1, n):
        s += 4 * f(a + i * h) if i % 2 != 0 else 2 * f(a + i * h)
    return s * h / 3.0

def simpson_38(f, a, b, n):
    h = (b - a) / n
    s = f(a) + f(b)
    for i in range(1, n):
        s += 2 * f(a + i * h) if i % 3 == 0 else 3 * f(a + i * h)
    return s * 3.0 * h / 8.0

def q1_compute_pi():
    a, b = 0.0, 1.0
    intervals = [6, 12, 24] # Initial, twice, four times
    true_pi = math.pi

    print("Approximating Pi = 4 * Integral(0 to 1) 1/(1+x^2) dx")
    print(f"{'Method':<16} | {'N':<4} | {'Computed Pi':<18} | {'Absolute Error'}")
    print("-" * 60)

    for n in intervals:
        # Multiply integral by 4 to compute Pi
        pi_trap = 4 * trapezoidal(f_q1_q2, a, b, n)
        pi_s13 = 4 * simpson_13(f_q1_q2, a, b, n)
        pi_s38 = 4 * simpson_38(f_q1_q2, a, b, n)

        print(f"{'Trapezoidal':<16} | {n:<4} | {pi_trap:<18.12f} | {abs(true_pi - pi_trap):.4e}")
        print(f"{'Simpson 1/3':<16} | {n:<4} | {pi_s13:<18.12f} | {abs(true_pi - pi_s13):.4e}")
        print(f"{'Simpson 3/8':<16} | {n:<4} | {pi_s38:<18.12f} | {abs(true_pi - pi_s38):.4e}")
        print("-" * 60)

q1_compute_pi()
```

Question 2 & Question 3 - Desired Accuracy Iteration

```python
def accuracy_iteration(f, a, b, true_val, target_err=0.5e-5, name="Integral"):
    print(f"\n--- {name} ---")
    print(f"{'N':<4} | {'Approximation':<18} | {'Absolute Error':<18} | {'Status'}")
    print("-" * 65)

    n = 2 # Smallest number of sub-intervals for Simpson's 1/3
    while True:
        approx = simpson_13(f, a, b, n)
        err = abs(approx - true_val)
        status = "Reached" if err < target_err else "Continuing..."

        print(f"{n:<4} | {approx:<18.12f} | {err:<18.4e} | {status}")

        if err < target_err:
            break
        n += 2 # Increase to next valid even number for Simpson's 1/3

def q2_q3_run():
    # Target Error
    tol = 0.5 * 10**(-5)

    # Q2: Integral of 1/(1+x^2) on [0, 1]
    accuracy_iteration(f_q1_q2, 0.0, 1.0, TRUE_VAL_Q1_Q2, tol, "Q2: f(x) = 1/(1+x^2) on [0, 1]")

    # Q3: Integral of sqrt(1-x^2) - x on [0, 1/sqrt(2)]
    upper_bound_q3 = 1.0 / math.sqrt(2.0)
    accuracy_iteration(f_q3, 0.0, upper_bound_q3, TRUE_VAL_Q3, tol, "Q3: f(x) = sqrt(1-x^2) - x on [0, 1/sqrt(2)]")

q2_q3_run()
```

Question 4 - Gauss-Legendre vs Newton-Cotes

```python
def gauss_legendre_2pt(f, a, b):
    # Roots: +/- 1/sqrt(3), Weights: 1, 1
    t1, t2 = -1.0/math.sqrt(3), 1.0/math.sqrt(3)
    w1, w2 = 1.0, 1.0

    # Transformation to [a, b]
    def x(t): return 0.5 * (b - a) * t + 0.5 * (b + a)
    dx_dt = 0.5 * (b - a)

    return dx_dt * (w1 * f(x(t1)) + w2 * f(x(t2)))

def gauss_legendre_3pt(f, a, b):
    # Roots: 0, +/- sqrt(3/5), Weights: 8/9, 5/9, 5/9
    t1, t2, t3 = 0.0, -math.sqrt(0.6), math.sqrt(0.6)
    w1, w2, w3 = 8.0/9.0, 5.0/9.0, 5.0/9.0

    def x(t): return 0.5 * (b - a) * t + 0.5 * (b + a)
    dx_dt = 0.5 * (b - a)

    return dx_dt * (w1 * f(x(t1)) + w2 * f(x(t2)) + w3 * f(x(t3)))

def q4_comparison():
    tasks = [
        ("Q2: 1/(1+x^2)", f_q1_q2, 0.0, 1.0, TRUE_VAL_Q1_Q2),
        ("Q3: sqrt(1-x^2) - x", f_q3, 0.0, 1.0/math.sqrt(2), TRUE_VAL_Q3)
    ]

    for name, f, a, b, true_val in tasks:
        print(f"\n--- {name} ---")
        print(f"{'Method (Evals)':<25} | {'Approximation':<18} | {'Absolute Error'}")
        print("-" * 65)

        # 2-Point Comparison
        gl2 = gauss_legendre_2pt(f, a, b)
        nc2 = trapezoidal(f, a, b, 1) # N=1 gives 2 evaluation points
        print(f"{'Gauss-Legendre 2-Pt':<25} | {gl2:<18.12f} | {abs(true_val - gl2):.4e}")
        print(f"{'Newton-Cotes 2-Pt (Trap)':<25} | {nc2:<18.12f} | {abs(true_val - nc2):.4e}")

        # 3-Point Comparison
        gl3 = gauss_legendre_3pt(f, a, b)
        nc3 = simpson_13(f, a, b, 2) # N=2 gives 3 evaluation points
        print(f"{'Gauss-Legendre 3-Pt':<25} | {gl3:<18.12f} | {abs(true_val - gl3):.4e}")
        print(f"{'Newton-Cotes 3-Pt (Simp)':<25} | {nc3:<18.12f} | {abs(true_val - nc3):.4e}")

q4_comparison()
```

Question 5 & Question 6 - Linear Systems

```python
def run_linear_solvers():
    # Initial Guesses
    TOL = 1e-3

    def iterate_jacobi(max_iter):
        x = [0.0, 0.0, 0.0, 0.0]
        for k in range(max_iter):
            x_new = [0.0] * 4
            x_new[0] = (-2 - x[1] + x[2] - x[3]) / 4.0
            x_new[1] = (-1 - x[0] + x[2] + x[3]) / 4.0
            x_new[2] = ( 0 + x[0] + x[1] - x[3]) / 5.0
            x_new[3] = ( 1 - x[0] + x[1] - x[2]) / 3.0

            # Step error (infinity norm)
            err = max(abs(x_new[i] - x[i]) for i in range(4))
            x = x_new

            # For Q6: Early stopping mechanism
            if max_iter > 20 and err < TOL:
                return x, k + 1, err
        return x, max_iter, err

    def iterate_gauss_seidel(max_iter):
        x = [0.0, 0.0, 0.0, 0.0]
        for k in range(max_iter):
            x_old = list(x)
            x[0] = (-2 - x[1] + x[2] - x[3]) / 4.0
            x[1] = (-1 - x[0] + x[2] + x[3]) / 4.0
            x[2] = ( 0 + x[0] + x[1] - x[3]) / 5.0
            x[3] = ( 1 - x[0] + x[1] - x[2]) / 3.0

            err = max(abs(x[i] - x_old[i]) for i in range(4))

            if max_iter > 20 and err < TOL:
                return x, k + 1, err
        return x, max_iter, err

    # Q5: Observe effect on error at k=10 and k=20 for Jacobi
    print("QUESTION 5: Jacobi Method at fixed Iterations")
    print(f"{'Iterations':<12} | {'Step Error (Max Diff)':<25} | {'Solution [x1, x2, x3, x4]'}")
    print("-" * 85)
    for k in [10, 20]:
        sol, iters, err = iterate_jacobi(k)
        formatted_sol = [round(val, 5) for val in sol]
        print(f"{k:<12} | {err:<25.4e} | {formatted_sol}")

    # Q6: Speed Comparison reaching TOL = 10^-3
    print("\nQUESTION 6: Convergence Speed Comparison to reach TOL = 1e-3")
    print(f"{'Method':<15} | {'Iterations required':<20} | {'Final Step Error'}")
    print("-" * 60)

    _, iters_jacobi, err_j = iterate_jacobi(100) # Arbitrary large number to allow break
    _, iters_gs, err_gs = iterate_gauss_seidel(100)

    print(f"{'Jacobi':<15} | {iters_jacobi:<20} | {err_j:.4e}")
    print(f"{'Gauss-Seidel':<15} | {iters_gs:<20} | {err_gs:.4e}")

run_linear_solvers()
```

Conclusion for Question 6:

The Gauss-Seidel method utilizes the most recently updated variable values immediately during execution. Because of this, it converged in just 7 iterations, compared to the Jacobi method which required 11 iterations to reach the same $10^{-3}$ tolerance step-error threshold. This confirms that Gauss-Seidel exhibits a notably faster speed of convergence for strictly and weakly diagonally dominant systems.
